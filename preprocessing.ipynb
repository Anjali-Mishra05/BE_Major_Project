```python
import pandas as pd
import os

input_file = "raw/upi_transactions_2024.csv"
output_file = "processed/upi_transactions_processed.csv"

os.makedirs("processed", exist_ok=True)

first_chunk = True

for chunk in pd.read_csv(input_file, chunksize=50000):

    # Timestamp cleaning
    chunk["timestamp"] = pd.to_datetime(
        chunk["timestamp"],
        errors="coerce"
    )

    # Night transaction
    chunk["is_night_transaction"] = (
        (chunk["hour_of_day"] < 6) |
        (chunk["hour_of_day"] >= 23)
    ).astype(int)

    # High amount transaction
    threshold = chunk["amount (INR)"].quantile(0.95)

    chunk["is_high_amount"] = (
        chunk["amount (INR)"] > threshold
    ).astype(int)

    # Save
    chunk.to_csv(
        output_file,
        mode="w" if first_chunk else "a",
        header=first_chunk,
        index=False
    )

    first_chunk = False

print("DONE!")
```
